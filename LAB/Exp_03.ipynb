# Experiment 3: Morphological Analysis Using NLTK


```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer

text = "The leaves are falling from trees easily while dogs were running."
print("Input Text:", text)
print("-" * 60)

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

tokens = word_tokenize(text)

print(f"{'Word':<12} | {'Stem (Porter)':<15} | {'Lemma (Noun)':<15} | {'Lemma (Verb)'}")
print("-" * 60)
for word in tokens:
    if word.isalpha():
        stem = stemmer.stem(word)
        lemma_n = lemmatizer.lemmatize(word.lower(), pos='n')
        lemma_v = lemmatizer.lemmatize(word.lower(), pos='v')
        print(f"{word:<12} | {stem:<15} | {lemma_n:<15} | {lemma_v}")
```
