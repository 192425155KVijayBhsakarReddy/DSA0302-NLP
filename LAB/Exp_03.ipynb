# Experiment 3: Morphological Analysis Using NLTK


```python
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer

text = "The leaves are falling from trees easily while dogs were running."
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
tokens = word_tokenize(text)

print(f"{'Word':<12} | {'Stem':<12} | {'Lemma(N)':<12} | {'Lemma(V)'}")
print("-" * 52)
for word in tokens:
    if word.isalpha():
        stem = stemmer.stem(word)
        ln = lemmatizer.lemmatize(word.lower(), pos='n')
        lv = lemmatizer.lemmatize(word.lower(), pos='v')
        print(f"{word:<12} | {stem:<12} | {ln:<12} | {lv}")
```
