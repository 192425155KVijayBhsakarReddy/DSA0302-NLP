# Experiment 6: Bigram Model for Text Generation


```python
import random
from collections import defaultdict

corpus = [
    "natural language processing is fun",
    "language processing enables artificial intelligence",
    "artificial intelligence is transforming the world",
    "natural language understanding is important"
]

# Build Bigrams
bigrams = defaultdict(list)
for sentence in corpus:
    words = sentence.split()
    for i in range(len(words) - 1):
        bigrams[words[i]].append(words[i + 1])

# Generate Text
random.seed(42)
start_word = "natural"
generated = [start_word]
current = start_word

for _ in range(8):
    next_words = bigrams.get(current)
    if not next_words:
        break
    current = random.choice(next_words)
    generated.append(current)

print("Training Corpus:")
for s in corpus:
    print(" -", s)
print("-" * 50)
print("Generated Text (Bigram Model):")
print(" ".join(generated))
```
