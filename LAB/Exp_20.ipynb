# Experiment 20: Information Retrieval System Using TF-IDF


```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

documents = [
    "Information retrieval is the science of searching for documents and metadata.",
    "Python is widely used for natural language processing and machine learning.",
    "Search engines use TF-IDF and relevance ranking algorithms.",
    "Deep learning models achieve high accuracy in NLP tasks."
]

query = ["natural language processing and search engines"]

# Vectorize documents and query
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(documents)
query_vec = vectorizer.transform(query)

# Compute Cosine Similarity
similarity_scores = cosine_similarity(query_vec, tfidf_matrix).flatten()

# Rank Documents
ranked_indices = similarity_scores.argsort()[::-1]

print("User Query:", query[0])
print("-" * 65)
print(f"{'Rank':<6} | {'Score':<8} | {'Document'}")
print("-" * 65)
for rank, idx in enumerate(ranked_indices, 1):
    print(f"{rank:<6} | {similarity_scores[idx]:<8.4f} | {documents[idx]}")
```
