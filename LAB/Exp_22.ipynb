# Experiment 22: Reference Resolution in Text


```python
# Pronominal reference resolution using recency and gender/number agreement
text = "John went to the market. He bought fresh apples. Mary met Alice and she smiled warmly."

sentences = [
    "John went to the market. He bought fresh apples.",
    "Mary met Alice and she smiled warmly.",
    "The laptop stopped working because it overheated."
]

def resolve_pronouns(sent):
    # Entity knowledge base with gender/type
    entities = {
        "John": "MALE",
        "Mary": "FEMALE",
        "Alice": "FEMALE",
        "laptop": "NEUTER"
    }
    pronoun_map = {
        "he": "MALE",
        "she": "FEMALE",
        "it": "NEUTER"
    }
    
    tokens = sent.replace(".", "").split()
    history = []
    resolutions = []
    
    for word in tokens:
        if word in entities:
            history.append((word, entities[word]))
        elif word.lower() in pronoun_map:
            gender = pronoun_map[word.lower()]
            # Find the most recent matching antecedent
            for ent, ent_gender in reversed(history):
                if ent_gender == gender:
                    resolutions.append((word, ent))
                    break
    return resolutions

print("Reference Resolution Analysis:")
print("-" * 60)
for s in sentences:
    res = resolve_pronouns(s)
    print(f"Sentence: \"{s}\"")
    for pronoun, antecedent in res:
        print(f"  -> Pronoun '{pronoun}' refers to antecedent: [{antecedent}]")
    print()
```
