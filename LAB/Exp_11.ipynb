# Experiment 11: Simple Top-Down Parser for CFG


```python
import nltk
from nltk import CFG, RecursiveDescentParser

# Define Context-Free Grammar
grammar = CFG.fromstring("""
  S -> NP VP
  NP -> Det N
  VP -> V NP
  Det -> 'the' | 'a'
  N -> 'dog' | 'cat'
  V -> 'chased' | 'saw'
""")

parser = RecursiveDescentParser(grammar)
sentence = "the dog chased a cat".split()

print("Grammar Rules:")
print(grammar)
print("-" * 45)
print("Parsing Sentence:", " ".join(sentence))
print("Parse Trees (Top-Down):")

trees = list(parser.parse(sentence))
for tree in trees:
    print(tree)
    tree.pretty_print()
```
