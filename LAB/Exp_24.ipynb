# Experiment 24: Dialog Act Recognition


```python
def recognize_dialog_act(u):
    u = u.strip().lower()
    if any(u.startswith(w) for w in ["hello", "hi", "hey"]):
        return "GREETING"
    if any(u.startswith(w) for w in ["bye", "goodbye"]):
        return "CLOSING"
    if any(u.startswith(w) for w in ["what", "where", "when", "why", "who", "how"]):
        return "WH_QUESTION"
    if any(u.startswith(w) for w in ["is", "are", "do", "does", "can"]) and u.endswith("?"):
        return "YES_NO_QUESTION"
    if any(u.startswith(w) for w in ["please", "turn", "open", "close"]):
        return "COMMAND"
    if any(u.startswith(w) for w in ["yes", "ok", "sure", "agreed"]):
        return "AGREEMENT"
    return "STATEMENT"

dialogue = [
    "Hello, how can I help you?",
    "Where is the nearest station?",
    "It is two blocks down.",
    "Can I walk there?",
    "Yes, absolutely.",
    "Please show me the route.",
    "Goodbye, have a nice day!"
]

for line in dialogue:
    print(f"{line:<32} : {recognize_dialog_act(line)}")
```
