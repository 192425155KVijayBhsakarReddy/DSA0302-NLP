# Experiment 24: Dialog Act Recognition


```python
def recognize_dialog_act(utterance):
    u = utterance.strip().lower()
    
    # Greetings
    if any(u.startswith(w) for w in ["hello", "hi", "hey", "good morning", "good evening"]):
        return "GREETING"
    
    # Closings
    if any(u.startswith(w) for w in ["bye", "goodbye", "see you", "have a nice day"]):
        return "CLOSING"
        
    # Wh-Questions
    if any(u.startswith(w) for w in ["what", "where", "when", "why", "who", "how"]):
        return "WH_QUESTION"
        
    # Yes/No Questions
    if any(u.startswith(w) for w in ["is", "are", "do", "does", "can", "could", "would", "will"]) and u.endswith("?"):
        return "YES_NO_QUESTION"
        
    # Directives / Commands
    if any(u.startswith(w) for w in ["please", "turn", "open", "close", "shut", "bring"]):
        return "DIRECTIVE / COMMAND"
        
    # Agreement / Acknowledgement
    if any(u.startswith(w) for w in ["yes", "yeah", "sure", "ok", "okay", "agreed"]):
        return "AGREEMENT"
        
    return "STATEMENT"

dialogue = [
    "Hello, how can I help you today?",
    "Where is the nearest train station?",
    "It is two blocks down to your left.",
    "Can I walk there in ten minutes?",
    "Yes, absolutely.",
    "Please show me the walking route.",
    "Goodbye, have a safe trip!"
]

print(f"{'Utterance':<40} | {'Recognized Dialog Act'}")
print("-" * 65)
for line in dialogue:
    act = recognize_dialog_act(line)
    print(f"{line:<40} | {act}")
```
