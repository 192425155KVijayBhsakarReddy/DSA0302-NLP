# Experiment 4: Finite-State Machine for Plural Nouns


```python
def noun_plural_fsm(noun):
    # Irregulars
    irregulars = {'child': 'children', 'man': 'men', 'foot': 'feet', 'tooth': 'teeth', 'mouse': 'mice', 'sheep': 'sheep'}
    w = noun.lower()
    if w in irregulars:
        return irregulars[w]
    
    # Sibilants: -s, -sh, -ch, -x, -z -> +es
    if w.endswith(('s', 'sh', 'ch', 'x', 'z')):
        return w + 'es'
    
    # Consonant + y -> -ies
    if w.endswith('y') and len(w) > 1 and w[-2] not in 'aeiou':
        return w[:-1] + 'ies'
        
    # -f or -fe -> -ves
    if w.endswith('fe'):
        return w[:-2] + 'ves'
    elif w.endswith('f') and not w.endswith(('ff', 'roof', 'cliff')):
        return w[:-1] + 'ves'
        
    # Default -> +s
    return w + 's'

nouns = ["cat", "dog", "box", "dish", "church", "baby", "boy", "leaf", "knife", "child", "man", "sheep"]

print(f"{'Singular':<12} | {'Plural'}")
print("-" * 30)
for n in nouns:
    print(f"{n:<12} | {noun_plural_fsm(n)}")
```
