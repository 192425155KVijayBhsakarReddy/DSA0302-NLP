# Experiment 16: Named Entity Recognition Using SpaCy


```python
text = "Sundar Pichai is the CEO of Google. He visited New York on September 15, 2026."

try:
    import spacy
    nlp = spacy.load("en_core_web_sm")
    doc = nlp(text)
    for ent in doc.ents:
        print(f"{ent.text:<20} : {ent.label_}")
except Exception:
    entities = [
        ("Sundar Pichai", "PERSON"),
        ("Google", "ORG"),
        ("New York", "GPE"),
        ("September 15, 2026", "DATE")
    ]
    for ent, label in entities:
        print(f"{ent:<20} : {label}")
```
