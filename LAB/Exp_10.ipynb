# Experiment 10: Transformation-Based Tagging


```python
# Step 1: Assign initial tags using default rule (most frequent tag = NN)
sentence = "The soldiers will race to the finish line"
tokens = sentence.split()

# Initial state: tag everything as NN except known function words
initial_tags = []
for w in tokens:
    if w.lower() in ["the", "a"]:
        initial_tags.append("DT")
    elif w.lower() in ["to"]:
        initial_tags.append("TO")
    elif w.lower() in ["will"]:
        initial_tags.append("MD")
    else:
        initial_tags.append("NN")

print("Input Sentence:", sentence)
print("-" * 45)
print("Initial Tagging:")
print(list(zip(tokens, initial_tags)))

# Step 2: Apply Transformation Rule
# Rule: If previous word is MD (modal) or TO, change tag from NN to VB
transformed_tags = list(initial_tags)
for i in range(1, len(transformed_tags)):
    if transformed_tags[i-1] in ["MD", "TO"] and transformed_tags[i] == "NN":
        print(f"-> Applying Rule: Change '{tokens[i]}' from NN to VB (preceded by {transformed_tags[i-1]})")
        transformed_tags[i] = "VB"

print("-" * 45)
print("Final Transformed Tagging:")
print(list(zip(tokens, transformed_tags)))
```
