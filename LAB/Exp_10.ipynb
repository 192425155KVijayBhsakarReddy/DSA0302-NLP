# Experiment 10: Transformation-Based Tagging


```python
sentence = "The soldiers will race to the finish line"
tokens = sentence.split()

tags = []
for w in tokens:
    if w.lower() in ["the", "a"]:
        tags.append("DT")
    elif w.lower() in ["to"]:
        tags.append("TO")
    elif w.lower() in ["will"]:
        tags.append("MD")
    else:
        tags.append("NN")

print("Initial Tags:", list(zip(tokens, tags)))

for i in range(1, len(tags)):
    if tags[i-1] in ["MD", "TO"] and tags[i] == "NN":
        tags[i] = "VB"

print("Transformed Tags:", list(zip(tokens, tags)))
```
