# Experiment 14: Sentence Agreement Checking with Feature Grammar


```python
from nltk import FeatureGrammar
from nltk.parse import FeatureChartParser

fgrammar = FeatureGrammar.fromstring("""
  S -> NP[NUM=?n] VP[NUM=?n]
  NP[NUM=?n] -> Det[NUM=?n] N[NUM=?n]
  VP[NUM=?n] -> V[NUM=?n] NP
  Det[NUM=sg] -> 'this' | 'the' | 'a'
  Det[NUM=pl] -> 'these' | 'the'
  N[NUM=sg] -> 'dog' | 'cat'
  N[NUM=pl] -> 'dogs' | 'cats'
  V[NUM=sg] -> 'runs' | 'chases'
  V[NUM=pl] -> 'run' | 'chase'
""")

parser = FeatureChartParser(fgrammar)
sentences = ["this dog runs", "these dogs run", "this dog run", "these dogs runs"]

for s in sentences:
    valid = bool(list(parser.parse(s.split())))
    print(f"{s:<18} : {'Valid (Agreement holds)' if valid else 'Invalid (Agreement error)'}")
```
