# Experiment 14: Sentence Agreement Checking with Feature Grammar


```python
import nltk
from nltk import FeatureGrammar
from nltk.parse import FeatureChartParser

# Feature Grammar with Number Agreement (sg = singular, pl = plural)
fgrammar = FeatureGrammar.fromstring("""
  S -> NP[NUM=?n] VP[NUM=?n]
  NP[NUM=?n] -> Det[NUM=?n] N[NUM=?n]
  VP[NUM=?n] -> V[NUM=?n] NP
  
  Det[NUM=sg] -> 'this' | 'the' | 'a'
  Det[NUM=pl] -> 'these' | 'the'
  
  N[NUM=sg] -> 'dog' | 'cat'
  N[NUM=pl] -> 'dogs' | 'cats'
  
  V[NUM=sg] -> 'runs' | 'chases'
  V[NUM=pl] -> 'run' | 'chase'
""")

parser = FeatureChartParser(fgrammar)

sentences = [
    "this dog runs",      # Valid: sg + sg
    "these dogs run",     # Valid: pl + pl
    "this dog run",       # Invalid: sg + pl
    "these dogs runs"     # Invalid: pl + sg
]

print("Agreement Checking Results:")
print("-" * 45)
for s in sentences:
    tokens = s.split()
    trees = list(parser.parse(tokens))
    status = "Grammatically Valid (Agreement holds)" if trees else "Grammatically Invalid (Agreement violation)"
    print(f"Sentence: '{s:<18}' -> {status}")
```
