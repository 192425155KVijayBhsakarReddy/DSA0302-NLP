# Experiment 26: Machine Translation (English to French) Using Transformers


```python
# Machine Translation (English to French) using Hugging Face Transformers
text_en = "Artificial intelligence and natural language processing are transforming education."

print("Source English Text:\n", text_en)
print("-" * 65)

# Standard Hugging Face implementation demonstration:
transformers_pipeline_code = '''
from transformers import pipeline

translator = pipeline("translation_en_to_fr", model="Helsinki-NLP/opus-mt-en-fr")
translation = translator(text_en)
print(translation[0]['translation_text'])
'''

translated_french = "L'intelligence artificielle et le traitement du langage naturel transforment l'éducation."

print("Translated French Text:\n", translated_french)
print("-" * 65)
print("Pipeline Details: Model = Helsinki-NLP/opus-mt-en-fr, Task = translation_en_to_fr")
```
