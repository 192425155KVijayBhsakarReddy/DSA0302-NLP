# Experiment 23: Text Coherence Evaluation


```python
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords

coherent_text = "Artificial intelligence is evolving. Machine learning models require datasets. These algorithms make predictions."
incoherent_text = "Artificial intelligence is evolving. Bananas are rich in potassium. The soccer match ended in shootout."

def coherence(text):
    sentences = [s for s in sent_tokenize(text) if s.strip()]
    sw = set(stopwords.words('english'))
    sent_words = [set(w.lower() for w in word_tokenize(s) if w.isalnum() and w.lower() not in sw) for s in sentences]
    scores = []
    for i in range(len(sent_words) - 1):
        s1, s2 = sent_words[i], sent_words[i+1]
        scores.append(len(s1 & s2) / len(s1 | s2) if (s1 | s2) else 0.0)
    return sum(scores) / len(scores) if scores else 0.0

print(f"Coherent Text Score   : {coherence(coherent_text):.4f}")
print(f"Incoherent Text Score : {coherence(incoherent_text):.4f}")
```
