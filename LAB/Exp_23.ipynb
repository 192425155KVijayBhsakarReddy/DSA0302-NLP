# Experiment 23: Text Coherence Evaluation


```python
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords

# Coherent vs Incoherent sample texts
coherent_text = """
Artificial intelligence is rapidly evolving. Machine learning models require large datasets for training.
These trained algorithms can recognize patterns and make accurate predictions.
"""

incoherent_text = """
Artificial intelligence is rapidly evolving. Bananas are rich in potassium and vitamins.
The soccer match ended in a dramatic penalty shootout.
"""

def calculate_coherence(text):
    sentences = [s for s in sent_tokenize(text) if s.strip()]
    if len(sentences) < 2:
        return 0.0
    
    stop_words = set(stopwords.words('english'))
    sentence_words = []
    for s in sentences:
        words = set(w.lower() for w in word_tokenize(s) if w.isalnum() and w.lower() not in stop_words)
        sentence_words.append(words)
        
    overlap_scores = []
    for i in range(len(sentence_words) - 1):
        s1, s2 = sentence_words[i], sentence_words[i+1]
        jaccard = len(s1 & s2) / len(s1 | s2) if (s1 | s2) else 0.0
        overlap_scores.append(jaccard)
        
    avg_coherence = sum(overlap_scores) / len(overlap_scores)
    return avg_coherence

print("Text Coherence Evaluation (Sentence-to-Sentence Lexical Overlap):")
print("-" * 65)
print(f"Coherent Text Score   : {calculate_coherence(coherent_text):.4f} (High Lexical Cohesion)")
print(f"Incoherent Text Score : {calculate_coherence(incoherent_text):.4f} (Low Lexical Cohesion)")
```
