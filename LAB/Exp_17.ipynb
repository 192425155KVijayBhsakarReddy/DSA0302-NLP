# Experiment 17: Access WordNet to Retrieve Synsets and Meanings


```python
from nltk.corpus import wordnet as wn

word = "bank"
synsets = wn.synsets(word)
print(f"Synsets for '{word}': {len(synsets)}")
for syn in synsets[:3]:
    print(f"{syn.name()} ({syn.pos()}): {syn.definition()}")
    print("Examples:", syn.examples())
    print("Lemmas:", [l.name() for l in syn.lemmas()])

s1 = wn.synset("dog.n.01")
s2 = wn.synset("cat.n.01")
print(f"Path Similarity between '{s1.name()}' and '{s2.name()}': {s1.path_similarity(s2):.4f}")
```
