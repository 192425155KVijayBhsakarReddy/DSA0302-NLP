# Experiment 17: Access WordNet to Retrieve Synsets and Meanings


```python
from nltk.corpus import wordnet as wn

word = "bank"
synsets = wn.synsets(word)

print(f"Synsets for '{word}': Total {len(synsets)} senses found.")
print("-" * 65)

for i, syn in enumerate(synsets[:5], 1):
    print(f"Sense {i}: {syn.name()} ({syn.pos()})")
    print(f"  Definition: {syn.definition()}")
    print(f"  Examples:   {syn.examples()}")
    print(f"  Lemmas:     {[l.name() for l in syn.lemmas()]}")
    print()

# Hypernyms & Similarity
syn1 = wn.synset("dog.n.01")
syn2 = wn.synset("cat.n.01")
print("Exploring Relationships:")
print(f"Dog Hypernyms: {[h.name() for h in syn1.hypernyms()]}")
print(f"Path Similarity between '{syn1.name()}' and '{syn2.name()}': {syn1.path_similarity(syn2):.4f}")
```
