# Experiment 2: Finite State Automaton (Strings Ending with 'ab')


```python
def fsa_ends_with_ab(string):
    state = 'q0'  # start state
    for ch in string:
        if state == 'q0':
            state = 'q1' if ch == 'a' else 'q0'
        elif state == 'q1':
            state = 'q1' if ch == 'a' else ('q2' if ch == 'b' else 'q0')
        elif state == 'q2':
            state = 'q1' if ch == 'a' else 'q0'
    return state == 'q2'

test_cases = ["ab", "aab", "bab", "baab", "aba", "b", "a", "abb"]

print(f"{'Input String':<15} | {'Result'}")
print("-" * 30)
for s in test_cases:
    result = "Accepted" if fsa_ends_with_ab(s) else "Rejected"
    print(f"{s:<15} | {result}")
```
