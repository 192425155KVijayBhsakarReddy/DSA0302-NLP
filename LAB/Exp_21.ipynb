# Experiment 21: Syntax-Driven Semantic Analysis (Noun Phrase Extraction)


```python
import nltk
from nltk import RegexpParser
from nltk.corpus import wordnet as wn

sentence = "The brilliant scientist presented an innovative theory at the grand conference."
tokens = nltk.word_tokenize(sentence)
pos_tags = nltk.pos_tag(tokens)

chunk_grammar = r"NP: {<DT>?<JJ>*<NN.*>+}"
chunk_parser = RegexpParser(chunk_grammar)
tree = chunk_parser.parse(pos_tags)

for subtree in tree.subtrees():
    if subtree.label() == 'NP':
        np_words = [w for w, t in subtree.leaves()]
        head = np_words[-1]
        syn = wn.synsets(head)
        defn = syn[0].definition() if syn else "N/A"
        print(f"NP: {' '.join(np_words):<28} | Head: {head:<12} | Meaning: {defn}")
```
