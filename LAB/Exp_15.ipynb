# Experiment 15: Probabilistic Context-Free Grammar (PCFG) Parsing


```python
from nltk import PCFG, ViterbiParser

pcfg_grammar = PCFG.fromstring("""
  S -> NP VP [1.0]
  NP -> Det N [0.7] | N [0.3]
  VP -> V NP [0.8] | V [0.2]
  Det -> 'the' [0.6] | 'a' [0.4]
  N -> 'astronomer' [0.5] | 'telescope' [0.5]
  V -> 'saw' [1.0]
""")

parser = ViterbiParser(pcfg_grammar)
sentence = "the astronomer saw a telescope".split()
for tree in parser.parse(sentence):
    print(tree)
    print(f"Probability: {tree.prob():.6f}")
    tree.pretty_print()
```
