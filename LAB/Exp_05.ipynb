# Experiment 5: Porter Stemmer Algorithm


```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
words = ["connection", "connections", "connective", "connecting", "connected", "troubled", "troubling", "troubles"]

print("Input Words:", words)
print("-" * 40)
print(f"{'Original Word':<18} | {'Stemmed Word'}")
print("-" * 40)
for word in words:
    print(f"{word:<18} | {stemmer.stem(word)}")
```
