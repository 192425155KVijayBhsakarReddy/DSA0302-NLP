# Experiment 19: Word Sense Disambiguation Using Lesk Algorithm


```python
from nltk.wsd import lesk
from nltk.tokenize import word_tokenize

sentences = [
    ("I went to the bank to deposit my paycheck.", "bank"),
    ("The fisherman sat on the river bank waiting for fish.", "bank")
]

for sent, word in sentences:
    tokens = word_tokenize(sent)
    sense = lesk(tokens, word)
    print(f"Sentence   : {sent}")
    print(f"Word       : {word}")
    print(f"Sense      : {sense.name()}")
    print(f"Definition : {sense.definition()}
")
```
