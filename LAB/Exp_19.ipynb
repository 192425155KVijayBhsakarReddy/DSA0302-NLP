# Experiment 19: Word Sense Disambiguation Using Lesk Algorithm


```python
from nltk.wsd import lesk
from nltk.tokenize import word_tokenize

sentences = [
    ("I went to the bank to deposit my paycheck.", "bank"),
    ("The fisherman sat on the river bank waiting for fish.", "bank"),
    ("He picked up a bat to hit the baseball.", "bat"),
    ("The vampire bat flew into the dark cave.", "bat")
]

print(f"{'Target Word':<12} | {'Context Sentence'}")
print("-" * 65)

for sent, word in sentences:
    tokens = word_tokenize(sent)
    best_sense = lesk(tokens, word)
    print(f"Sentence   : \"{sent}\"")
    print(f"Disambiguated Sense : {best_sense.name()}")
    print(f"Definition : {best_sense.definition()}")
    print("-" * 65)
```
