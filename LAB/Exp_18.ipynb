# Experiment 18: First-Order Predicate Calculus (FOPC) Parser


```python
from nltk.sem import logic

# Initialize logic expression reader
read_expr = logic.Expression.fromstring

expressions = [
    "man(socrates)",
    "mortal(socrates)",
    "all x. (man(x) -> mortal(x))",
    "exists x. (dog(x) & friendly(x))",
    "loves(john, mary) | loves(mary, john)"
]

print("First-Order Logic Parsing:")
print("-" * 50)
for expr_str in expressions:
    parsed = read_expr(expr_str)
    print(f"Input Logic String : {expr_str}")
    print(f"Parsed Expression  : {parsed}")
    print(f"Type / Class       : {type(parsed).__name__}")
    print(f"Free Variables     : {parsed.free()}")
    print("-" * 50)
```
