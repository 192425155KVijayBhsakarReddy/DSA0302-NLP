# Experiment 13: Generate Parse Tree Using CFG


```python
from nltk import CFG, ChartParser

grammar = CFG.fromstring("""
  S -> NP VP
  NP -> Det Adj N | Det N
  VP -> V NP | V
  Det -> 'the' | 'a'
  Adj -> 'quick' | 'brown'
  N -> 'fox' | 'dog'
  V -> 'jumps' | 'sees'
""")

parser = ChartParser(grammar)
sentence = "the quick fox sees a dog".split()
for i, t in enumerate(parser.parse(sentence), 1):
    print(f"Parse Tree {i}:")
    print(t)
    t.pretty_print()
```
