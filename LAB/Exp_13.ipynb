# Experiment 13: Generate Parse Tree Using CFG


```python
import nltk
from nltk import CFG, ChartParser

grammar = CFG.fromstring("""
  S -> NP VP
  NP -> Det Adj N | Det N
  VP -> V NP | V
  Det -> 'the' | 'a'
  Adj -> 'quick' | 'brown'
  N -> 'fox' | 'dog'
  V -> 'jumps' | 'sees'
""")

parser = ChartParser(grammar)
sentence = "the quick fox sees a dog".split()

print("Sentence:", " ".join(sentence))
print("-" * 45)
trees = list(parser.parse(sentence))
for i, t in enumerate(trees, 1):
    print(f"Parse Tree {i}:\n")
    print(t)
    print("\nGraphical Structure:")
    t.pretty_print()
```
