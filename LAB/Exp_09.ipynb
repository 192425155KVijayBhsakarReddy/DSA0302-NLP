# Experiment 9: Rule-Based POS Tagging Using Regular Expressions


```python
import nltk
from nltk.tag import RegexpTagger

# Regular expression rules for POS tagging
patterns = [
    (r'.*ing$', 'VBG'),               # gerunds
    (r'.*ed$', 'VBD'),                # simple past
    (r'.*es$', 'VBZ'),                # 3rd singular present
    (r'.*ould$', 'MD'),               # modals
    (r'.*\'s$', 'NN$'),               # possessive nouns
    (r'.*s$', 'NNS'),                 # plural nouns
    (r'^-?[0-9]+(\.[0-9]+)?$', 'CD'), # cardinal numbers
    (r'.*ly$', 'RB'),                 # adverbs
    (r'.*able$', 'JJ'),               # adjectives
    (r'^(the|a|an)$', 'DT'),          # determiners
    (r'.*', 'NN')                     # nouns (default)
]

tagger = RegexpTagger(patterns)

sentence = "The 3 playing cats quickly jumped over the fence"
tokens = sentence.split()
tagged = tagger.tag(tokens)

print("Input Sentence:", sentence)
print("-" * 35)
print(f"{'Word':<12} | {'Assigned POS Tag'}")
print("-" * 35)
for word, tag in tagged:
    print(f"{word:<12} | {tag}")
```
