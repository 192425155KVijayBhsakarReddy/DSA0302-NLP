# Experiment 9: Rule-Based POS Tagging Using Regular Expressions


```python
from nltk.tag import RegexpTagger

patterns = [
    (r'.*ing$', 'VBG'),
    (r'.*ed$', 'VBD'),
    (r'.*es$', 'VBZ'),
    (r'.*ould$', 'MD'),
    (r'.*\'s$', 'NN$'),
    (r'.*s$', 'NNS'),
    (r'^-?[0-9]+(\.[0-9]+)?$', 'CD'),
    (r'.*ly$', 'RB'),
    (r'.*able$', 'JJ'),
    (r'^(the|a|an)$', 'DT'),
    (r'.*', 'NN')
]

tagger = RegexpTagger(patterns)
sentence = "The 3 playing cats quickly jumped over the fence"
for word, tag in tagger.tag(sentence.split()):
    print(f"{word:<10} : {tag}")
```
