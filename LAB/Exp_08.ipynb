# Experiment 8: Stochastic Part-of-Speech Tagging


```python
from collections import defaultdict

tagged_corpus = [
    [("the", "DT"), ("cat", "NN"), ("sat", "VBD"), ("on", "IN"), ("the", "DT"), ("mat", "NN")],
    [("a", "DT"), ("dog", "NN"), ("barked", "VBD"), ("at", "IN"), ("the", "DT"), ("cat", "NN")],
    [("the", "DT"), ("dog", "NN"), ("sat", "VBD"), ("on", "IN"), ("a", "DT"), ("rug", "NN")]
]

word_tag_freq = defaultdict(lambda: defaultdict(int))
for sentence in tagged_corpus:
    for word, tag in sentence:
        word_tag_freq[word.lower()][tag] += 1

def stochastic_pos_tagger(sentence):
    tokens = sentence.lower().split()
    tagged = []
    for token in tokens:
        best_tag = max(word_tag_freq[token], key=word_tag_freq[token].get) if token in word_tag_freq else "NN"
        tagged.append((token, best_tag))
    return tagged

test_sentence = "The dog sat on the rug"
for word, tag in stochastic_pos_tagger(test_sentence):
    print(f"{word:<10} : {tag}")
```
