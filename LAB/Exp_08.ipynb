# Experiment 8: Stochastic Part-of-Speech Tagging


```python
from collections import defaultdict

# Simple training dataset: list of (word, tag)
tagged_corpus = [
    [("the", "DT"), ("cat", "NN"), ("sat", "VBD"), ("on", "IN"), ("the", "DT"), ("mat", "NN")],
    [("a", "DT"), ("dog", "NN"), ("barked", "VBD"), ("at", "IN"), ("the", "DT"), ("cat", "NN")],
    [("the", "DT"), ("dog", "NN"), ("sat", "VBD"), ("on", "IN"), ("a", "DT"), ("rug", "NN")]
]

# Calculate word-tag frequencies (Emission counts)
word_tag_freq = defaultdict(lambda: defaultdict(int))
for sentence in tagged_corpus:
    for word, tag in sentence:
        word_tag_freq[word.lower()][tag] += 1

def stochastic_pos_tagger(sentence):
    tokens = sentence.lower().split()
    tagged_sentence = []
    for token in tokens:
        if token in word_tag_freq:
            # Assign tag with highest probability (argmax)
            best_tag = max(word_tag_freq[token], key=word_tag_freq[token].get)
        else:
            best_tag = "NN"  # default fallback
        tagged_sentence.append((token, best_tag))
    return tagged_sentence

test_sentence = "The dog sat on the rug"
result = stochastic_pos_tagger(test_sentence)

print("Input Sentence:", test_sentence)
print("-" * 35)
print(f"{'Word':<12} | {'Assigned POS Tag'}")
print("-" * 35)
for word, tag in result:
    print(f"{word:<12} | {tag}")
```
