```python
def doublable(word):
    vowels = "aeiou"

    if len(word) < 3:
        return False

    if word[-1] in "wxy":
        return False

    if not (word[-3] not in vowels and
            word[-2] in vowels and
            word[-1] not in vowels):
        return False

    groups = 0
    in_vowel = False

    for ch in word:
        if ch in vowels:
            if not in_vowel:
                groups += 1
            in_vowel = True
        else:
            in_vowel = False

    return groups == 1


def inflect(word, feature):
    if feature in ("N+PL", "V+3SG"):
        if word.endswith(("s", "x", "z", "ch", "sh")):
            return word + "es"
        if len(word) > 1 and word.endswith("y") and word[-2] not in "aeiou":
            return word[:-1] + "ies"
        if len(word) > 1 and word.endswith("o") and word[-2] not in "aeiou":
            return word + "es"
        return word + "s"

    if feature == "V+PROG":
        if len(word) >= 3 and word.endswith("e") and word[-2] not in "aeiou":
            return word[:-1] + "ing"
        if doublable(word):
            return word + word[-1] + "ing"
        return word + "ing"

    suffix = {"V+PAST": "ed", "ADJ+COMP": "er", "ADJ+SUP": "est"}[feature]

    if word.endswith("e"):
        return word + {"V+PAST": "d", "ADJ+COMP": "r", "ADJ+SUP": "st"}[feature]

    if len(word) > 1 and word.endswith("y") and word[-2] not in "aeiou":
        return word[:-1] + "i" + suffix

    if doublable(word):
        return word + word[-1] + suffix

    return word + suffix


L = int(input())
lexicon = {}

for _ in range(L):
    stem, feature, form = input().split()
    lexicon[(stem, feature)] = form

Q = int(input())

for _ in range(Q):
    stem, feature = input().split()
    if (stem, feature) in lexicon:
        print(lexicon[(stem, feature)])
    else:
        print(inflect(stem, feature))
```
