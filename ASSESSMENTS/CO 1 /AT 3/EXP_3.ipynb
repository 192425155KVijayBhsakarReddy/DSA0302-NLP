```python
def double(w):
    return (len(w)>=3 and w[-3] not in "aeiou" and
            w[-2] in "aeiou" and w[-1] not in "aeiouwxy" and
            sum(c in "aeiou" and (i==0 or w[i-1] not in "aeiou")
                for i,c in enumerate(w))==1)

L=int(input())
d={}
for _ in range(L):
    w,f,v=input().split()
    d[w,f]=v

def inf(w,f):
    if f in ("N+PL","V+3SG"):
        if w.endswith(("s","x","z","ch","sh")): return w+"es"
        if w.endswith("y") and w[-2] not in "aeiou": return w[:-1]+"ies"
        if w.endswith("o") and len(w)>1 and w[-2] not in "aeiou": return w+"es"
        return w+"s"
    if f=="V+PROG":
        if len(w)>=3 and w.endswith("e") and w[-2] not in "aeiou": return w[:-1]+"ing"
        return w+w[-1]+"ing" if double(w) else w+"ing"
    s={"V+PAST":"ed","ADJ+COMP":"er","ADJ+SUP":"est"}[f]
    if w.endswith("e"): return w+{"ed":"d","er":"r","est":"st"}[s]
    if w.endswith("y") and w[-2] not in "aeiou": return w[:-1]+"i"+s
    return w+w[-1]+s if double(w) else w+s

for _ in range(int(input())):
    w,f=input().split()
    print(d.get((w,f),inf(w,f)))
```
