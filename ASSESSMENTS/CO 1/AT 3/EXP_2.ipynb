```python
n, k = map(int, input().split())
alphabet = input().split()
start = int(input())

data = list(map(int, input().split()))
accepting = set(data[1:])

t = int(input())
transitions = {}

for _ in range(t):
    p, c, r = input().split()
    p, r = int(p), int(r)

    if (p, c) not in transitions:
        transitions[(p, c)] = set()

    transitions[(p, c)].add(r)

q = int(input())

for _ in range(q):
    word = input()
    states = {start}

    if word == "#":
        word = ""

    for c in word:
        next_states = set()

        for state in states:
            next_states.update(transitions.get((state, c), set()))

        states = next_states

    if states & accepting:
        print("ACCEPT")
    else:
        print("REJECT")
```
