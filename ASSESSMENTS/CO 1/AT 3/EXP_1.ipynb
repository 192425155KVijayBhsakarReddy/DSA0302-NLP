```python
import re

pattern = r"[A-Za-z]+(?:['-][A-Za-z]+)*"

n = int(input())

for _ in range(n):
    line = input()
    tokens = re.findall(pattern, line)

    if tokens:
        print(len(tokens), *tokens)
    else:
        print(0)
```
